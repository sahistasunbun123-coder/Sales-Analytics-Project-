```python
import pandas as pd

df = pd.read_csv('/content/retail_sales_dataset.csv')
df['transaction_date'] = pd.to_datetime(df['transaction_date'], errors='coerce')

print(df['transaction_date'].isnull().sum())
```

```python
print(df['category'].nunique())
```

```python
print(((df['discount_pct'] < 0) | (df['discount_pct'] > 100)).sum())
```

```python
expected = df['quantity'] * df['unit_price'] * (1 - df['discount_pct']/100)
print(((df['sales_amount'] - expected).abs() > 0.01).sum())
```

```python
print("Quantity:", (df['quantity'] <= 0).sum())
print("Unit Price:", (df['unit_price'] <= 0).sum())
print("Sales:", (df['sales_amount'] <= 0).sum())
```

```python
for c in ['quantity','unit_price','sales_amount']:
    z = ((df[c]-df[c].mean())/df[c].std()).abs()
    print(c, "outliers:", (z > 3).sum())
```

```python
print("Missing Customer IDs:", df['customer_id'].isnull().sum())
print("Total Customers:", df['customer_id'].unique())
```

```python
print(df.groupby('product_id')[['product_name','category','brand']].nunique().gt(1).any(axis=1).sum())
```

```python
print(df['region'].nunique())
```

```python
print("Total Revenue:", df['sales_amount'].sum())
```

```python
print("Total Orders:", df['transaction_id'].nunique())
```

```python
print("Total Customers:", df['customer_id'].nunique())
```

```python
print("Average Order Value:", df['sales_amount'].mean())
```

```python
print("Units Sold:", df['quantity'].sum())
```

```python
df['month'] = df['transaction_date'].dt.to_period('M')
monthly = df.groupby('month')['sales_amount'].sum()
print(monthly)
```

```python
growth = ((monthly.iloc[-1] - monthly.iloc[0]) / monthly.iloc[0]) * 100
print("Revenue Growth %:", growth)
```

```python
region_sales = df.groupby('region')['sales_amount'].sum()
print(region_sales)
```

```python
category_sales = df.groupby('category')['sales_amount'].sum()
print(category_sales)
```

```python
top10 = df.groupby('product_name')['sales_amount'].sum().sort_values(ascending=False).head(10)
print(top10)
```

```python
print(((df['quantity']*df['unit_price']).sum()-df['sales_amount'].sum())/(df['quantity']*df['unit_price']).sum()*100)
```
